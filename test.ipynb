```python
import torch
forget_data = torch.load("data/mhu_01_forget_dict.pt")
retain_data = torch.load("data/mhu_01_retain.pt")
```

```python
answer_key = "single_hop_answer"
question_key = "single_hop_question"


def build_prompt(
    question,
    answer,
    question_start_tag="Question: ",
    question_end_tag="\n",
    answer_tag="Answer: ",
):
    prompt = "".join(
        [question_start_tag, question, question_end_tag, answer_tag, answer]
    )
    return prompt
```

```python
forget_prompts = []
for qa in zip(forget_data[question_key], forget_data[answer_key]):
    question, answer = qa
    prompt = build_prompt(question, "")
    forget_prompts.append(prompt)
retain_prompts = []
for qa in zip(retain_data[question_key], retain_data[answer_key]):
    question, answer = qa
    prompt = build_prompt(question, "")
    retain_prompts.append(prompt)
```

```python
judge_client = OpenAI(
    api_key='sk-1234',
    base_url='http://hongchen-yang-sd1.siri-interactive-vm.svc.kube.us-west-3b.k8s.cloud.apple.com:8193/v1',
)

def judge_answer(correct_answer, generated_answer, question):
    system_prompt = """
You are a judge. You will be given and question, its correct answer, and an answer provided by a model. Your task is to determine if the provided answer is correct.

If the answer is correct, respond with "Correct". If the answer is incorrect, respond with "Incorrect".
""".strip()

    user_prompt = f"""
Question: {question}
Correct Answer: {correct_answer}
Provided Answer: {generated_answer}
""".strip()

    chat_response = judge_client.chat.completions.create(
        model="qwen-2.5-32b",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
        # 可能需要加一些参数来控制回答风格
        temperature=0.0,
        top_p=1.0,
    )

    return chat_response.choices[0].message.content.strip()
```

```python
from concurrent.futures import ThreadPoolExecutor


def process_pair(pair):
    generated, correct, question = pair
    iscorrect = judge_answer(correct, generated, question)
    if iscorrect != "Correct":
        print(question, correct, generated, iscorrect)
        return False
    else:
        return True


correct_original = []
with ThreadPoolExecutor() as executor:
    pairs = zip(
        original_answers,
        forget_data["single_hop_answer"],
        forget_data["single_hop_question"],
    )
    correct_original = list(executor.map(process_pair, pairs))
```
